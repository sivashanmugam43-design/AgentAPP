# Translate missing English (free T4 GPU)

1. **Runtime → Change runtime type → T4 GPU.**
2. Put `temples.jsonl` (from `output/` on your laptop) in Google Drive at `MyDrive/AgentAPP/temples.jsonl`.
3. IndicTrans2 is gated: accept its terms at https://huggingface.co/ai4bharat/indictrans2-indic-en-dist-200M,
   create a token at https://huggingface.co/settings/tokens, and add it in Colab's **Secrets** panel (key icon)
   as `HF_TOKEN` with notebook access on. Without it, the script falls back to NLLB.
4. Run the cell. The translated file and the cache are saved to Drive as it goes, so if the session
   disconnects just run the cell again: it resumes.
5. Download `temples.jsonl` back into `output/` and run `python main.py --stage export` for the CSV/XLSX.
